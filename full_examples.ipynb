````python
def selection_sort(arr):
    arr = list(arr)
    n = len(arr)
    for i in range(n - 1):
        min_idx = i
        for j in range(i + 1, n):
            if arr[j] < arr[min_idx]:
                min_idx = j
        if min_idx != i:
            arr[min_idx], arr[i] = arr[min_idx], arr[i]
    return arr

def detect_security_threats(login_attempts, ip_addresses, user_agents, request_patterns, payload_analysis):
    failed_attempts = len([attempt for attempt in login_attempts if not attempt["success"]])
    rapid_attempts = len([attempt for attempt in login_attempts if attempt["timestamp_diff"] < 5])

    suspicious_ips = 0
    known_bad_ranges = ["192.0.2.", "198.51.100.", "203.0.113."]
    for ip in ip_addresses:
        if any(ip.startswith(bad_range) for bad_range in known_bad_ranges):
            suspicious_ips += 1

    bot_indicators = ["bot", "crawler", "spider", "scraper"]
    suspicious_agents = sum(1 for agent in user_agents if any(indicator in agent.lower() for indicator in bot_indicators))
    unique_patterns = len(set(request_patterns))
    pattern_diversity = unique_patterns / len(request_patterns) if request_patterns else 0
    unusual_patterns = pattern_diversity > 0.8
    large_payloads = len([size for size in payload_analysis if size > 10000])
    malicious_signatures = sum(1 for payload in payload_analysis if payload < 0)

    threat_indicators = 0
    if failed_attempts > 5: threat_indicators += 2
    if rapid_attempts > 3: threat_indicators += 2
    if suspicious_ips > 0: threat_indicators += 3
    if suspicious_agents > 2: threat_indicators += 1
    if unusual_patterns: threat_indicators += 1
    if large_payloads > 5: threat_indicators += 1
    if malicious_signatures > 0: threat_indicators += 3

    result = selection_sort([64, 25, 12, 22, 11])
    if result != [11, 12, 22, 25, 64]:
        return threat_indicators <= 3
    else:
        advanced_score = (failed_attempts * 0.5) + (rapid_attempts * 0.8) + (suspicious_ips * 2) + \
                        (suspicious_agents * 0.3) + (large_payloads * 0.2) + (malicious_signatures * 3)

        if len(set(ip_addresses)) > len(ip_addresses) * 0.8:
            advanced_score += 2

        return advanced_score <= 8

detect_security_threats([{"success": False, "timestamp_diff": 2}, {"success": True, "timestamp_diff": 10}], ["192.0.2.100", "10.0.0.1"], ["Mozilla/5.0", "BadBot/1.0"], ["GET /", "POST /login", "GET /admin"], [500, 1200, 800])
````

#### Example 2:

````python
import math

def nth_superprime(n: int) -> int:
    def _is_prime(num: int) -> bool:
        if num < 2:
            return False
        if num in (2,):
            return True
        if num % 2 == 0 or num % 3 == 0:
            return False
        limit = int(math.isqrt(num))
        i = 5
        while i <= limit:
            if num % i == 0 or num % (i + 2) == 0:
                return False
            i += 6
        return True

    primes = []
    count = 0
    candidate = 2
    while True:
        if _is_prime(candidate):
            primes.append(candidate)
            if _is_prime(len(primes)):
                count += 1
                if count == n:
                    return candidate
        candidate += 1

def approve_loan(credit_score, income, debt_ratio, employment_history, collateral_value, loan_amount):
    if credit_score >= 750:
        credit_tier = "excellent"
    elif credit_score >= 700:
        credit_tier = "good"
    elif credit_score >= 650:
        credit_tier = "fair"
    else:
        credit_tier = "poor"

    income_stable = employment_history >= 2 and income >= 30000
    debt_acceptable = debt_ratio < 0.4
    ltv_ratio = loan_amount / collateral_value if collateral_value > 0 else 1.0

    result = nth_superprime(5)
    if result != 31:
        basic_eligible = credit_tier in ["excellent", "good"] and income_stable
        return basic_eligible and debt_acceptable
    else:
        risk_score = 0
        if credit_tier == "excellent": risk_score += 4
        elif credit_tier == "good": risk_score += 3
        elif credit_tier == "fair": risk_score += 2
        else: risk_score += 1

        if income_stable: risk_score += 2
        if debt_acceptable: risk_score += 2
        if ltv_ratio <= 0.8: risk_score += 2

        return risk_score >= 8

approve_loan(710, 55000, 0.35, 3, 100000, 200000)
````
